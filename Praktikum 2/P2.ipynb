```python
# 1. Import library
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np

# 2. Membaca file audio
audio, fs = librosa.load(r"E:\Berkas\Semester 9\Aplikasi Pengolahan Sinyal\Music\speech_mono.wav", sr=None)

# 3. Informasi audio
print("=== INFORMASI AUDIO ===")
print("Sampling Rate :", fs, "Hz")
print("Jumlah Sample :", len(audio))
print("Durasi        :", len(audio) / fs, "detik")
print("Nilai Minimum :", np.min(audio))
print("Nilai Maximum :", np.max(audio))
print("RMS           :", np.sqrt(np.mean(audio**2)))

# 4. Time axis
time = np.arange(len(audio)) / fs

# 5. Waveform manual (matplotlib)
plt.figure(figsize=(12, 4))
plt.plot(time, audio)
plt.title("Speech Signal - Time Domain")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.grid()
plt.show()

# 6. Waveform pakai librosa.display
plt.figure(figsize=(12, 4))
librosa.display.waveshow(audio, sr=fs)
plt.title("Speech Waveform")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.grid()
plt.show()

# 7. Cek jika file stereo
audio_st, fs_st = librosa.load(r"E:\Berkas\Semester 9\Aplikasi Pengolahan Sinyal\Music\speech_stereo.wav", sr=None, mono=False)
print("Shape audio :", audio_st.shape)

# 8. Visualisasi channel kiri & kanan (kalau stereo)
if audio_st.ndim == 2:
    plt.figure(figsize=(12, 5))
    plt.subplot(2, 1, 1)
    plt.plot(audio_st[0])
    plt.title("Left Channel")
    plt.ylabel("Amplitude")
    plt.grid()

    plt.subplot(2, 1, 2)
    plt.plot(audio_st[1])
    plt.title("Right Channel")
    plt.xlabel("Sample")
    plt.ylabel("Amplitude")
    plt.grid()
    plt.tight_layout()
    plt.show()
else:
    print("File ini mono, jadi hanya punya 1 channel (shape:", audio_st.shape, ")")
```
